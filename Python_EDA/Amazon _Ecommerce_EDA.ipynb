```python
# =================================
#   (Product wise Total_Sales)
# =================================
# Filter only delivered orders
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

# Calculate product-wise total sales
product_sales = delivered_sales.groupby('Product_Name')['Total_Sales_INR'].sum()

# Create Bar Chart
product_sales.plot(kind='bar', figsize=(12,6))

# Add title and labels
plt.title('Product-wise Delivered Sales')
plt.xlabel('Product Name')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

delivered_sales.groupby('Product_Name')['Total_Sales_INR'].sum()
```

```python
# ==========================================
#   (Month wise Total Sales)
# ==========================================

# Filter only Delivered orders
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

# Calculate Month-wise Total Delivered Sales
monthly_sales = delivered_sales.groupby('Month_Name')['Total_Sales_INR'].sum()

# Arrange months in calendar order
month_order = ['January', 'February', 'March', 'April',
               'May', 'June', 'July', 'August',
               'September', 'October', 'November', 'December']

monthly_sales = monthly_sales.reindex(month_order).dropna()

# Create Bar Chart
monthly_sales.plot(kind='bar', figsize=(12,6), color='skyblue')

plt.title('Month-wise Total Delivered Sales', fontsize=14)
plt.xlabel('Month Name')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

delivered_sales.groupby('Month_Name')['Total_Sales_INR'].sum().sort_values(ascending=False)
```

```python
# ======================================
#  (State wise Total Sales)
# =====================================

# Filter only Delivered orders
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

# Calculate State-wise Total Sales
state_sales = delivered_sales.groupby('State')['Total_Sales_INR'].sum().sort_values(ascending=False)

# Create Bar Graph
state_sales.plot(kind='bar', figsize=(14,6), color='orange')

plt.title('State-wise Total Delivered Sales')
plt.xlabel('State')
plt.ylabel('Total Sales (INR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delivered_sales = df[df['Delivery_Status'] == 'Delivered']

delivered_sales.groupby('State')['Total_Sales_INR'].sum().sort_values(ascending=False)
```

```python
# ================================================
#  (Delivery Status-wise Total Order Value (INR))
# ================================================

delivery_sales = df.groupby('Delivery_Status')['Total_Sales_INR'].sum()

plt.figure(figsize=(8, 6))

plt.pie(delivery_sales,
        labels=delivery_sales.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Delivery Status Wise Order Amount')
plt.show()
```
